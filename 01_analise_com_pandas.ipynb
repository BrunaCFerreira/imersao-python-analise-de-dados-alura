Importação

tem que importar a biblioteca pandas no começo do código pq ela ajuda a organizar, manipular e analisar dados em tabelas. (pd) é o apelido que demos ao pandas

```python
import pandas as pd
import plotly.express as px
```

df é só um nome que a gente costuma usar para variáveis que guardam DataFrames (tabelas do pandas). Aí posso colocar df_ e o nome que eu quiser.

Quando escrevo df_principal embaixo, estou mostrando o conteúdo da variável que criei em cima.

sheet_name é para dizer o nome da aba da planilha que eu tô chamando.

pd é a abreviação de pandas e .read_excel é a função que lê um arquivo Excel.

```python
df_principal = pd.read_excel("content/Atividade prática Alura.xlsx", sheet_name="Principal")
df_principal
```

como a planilha tem mais de uma aba, copiei e colei oq já havia feito e editei o nome da váriavel e o nome da aba no sheet_name

```python
df_Total_de_acoes = pd.read_excel("content/Atividade prática Alura.xlsx", sheet_name="Total_de_acoes")
df_Total_de_acoes
```

```python
df_Ticker = pd.read_excel("content/Atividade prática Alura.xlsx", sheet_name="Ticker")
df_Ticker
```

```python
df_CHATGPT = pd.read_excel("content/Atividade prática Alura.xlsx", sheet_name="CHATGPT")
df_CHATGPT
```

Tem algumas colunas da planilha que não vão ser necessárias, então vamos selecionar só as que queremos usar. Entre os [ ] duplos se coloca as que quer MANTER, entre aspas e separadas por vírgula. Isso não apaga as colunas da planilha original.

O .copy serve para criar uma cópia dos dados selecionados, sem alterar o DataFrame original.

```python
df_principal = df_principal[['Ativo', 'Data', 'Último (R$)', 'Var. Dia (%)']].copy()
df_principal
```

No Python dá para usar nomes de colunas com acentos e caracteres especiais, mas às vezes é mais prático simplificar para evitar b.o em algumas operações. Para isso usamos .rename(columns={ }). Aí dentro a gente bota o nome atual entre aspas, dois-pontos e o nome novo entre aspas também. Se for mudar mais de um, separa por vírgula.

O .copy no final cria uma cópia independente; não apaga o original. Como atribuímos o resultado a df_principal, essa variável passa a apontar para a versão renomeada.

```python
df_principal = df_principal.rename(columns = {'Último (R$)':'valor_final','Var. Dia (%)':'var_dia_pct'}).copy()
df_principal
```

o ['Var_pct'] é a nova coluna que eu to criando ai eu atribui com o = que dentro dela eu vou estar colocando o resultado da divisão da coluna ['var_dia_pct'] por 100

Ai para descobrir o valor inicial (a gnt criou uma nova coluna pra ele) pegamos o valor final e dividimos pelo var_pct que acabamos de descobrir + 1

obs: tem que chamar o principal pq é nela que ta essa coluna




```python
df_principal['Var_pct'] = df_principal['var_dia_pct'] / 100
df_principal['valor_inicial'] = df_principal['valor_final'] / (df_principal['Var_pct'] + 1)
df_principal
```

merge é o código usado para juntar tabelas usando valores em comum, tipo uma ideia parecida com o PROCV do Excel (mas não é a mesma coisa).

O df_Total_de_acoes dentro do ( ) é a outra tabela que queremos juntar à principal.

left_on diz qual coluna vamos usar da tabela da esquerda (Ativo) e right_on diz a coluna da tabela da direita (Código).

how="left" significa que vamos manter as linhas da tabela da esquerda, mesmo quando não houver correspondência na outra.

Ou seja, vamos associar os ativos às quantidades teóricas correspondentes.

```python
df_principal = df_principal.merge(df_Total_de_acoes, left_on='Ativo', right_on='Código', how='left')
df_principal
```

Como usar código e ativo ficou repetitivo e desnecessário vamos dropar a coluna "código' e para isso usamos o código .drop

```python
df_principal = df_principal.drop(columns=['Código'])
df_principal
```

Aqui a gente vai criar uma nova coluna que vai ser a Variacao_rs e nela vamos botar o seguinte calculo :(valor_final	- valor_inicial) * Qtde. Teórica

```python
df_principal['Variacao_rs'] = (df_principal['valor_final'] - df_principal['valor_inicial'])*df_principal['Qtde. Teórica']
df_principal
```

alterando a maneira como os números grandes nas colunas aparecem em float (tipo de dados que armazenam dados quebrados)

```python
pd.options.display.float_format = '{:.2f}'.format
```

A coluna Qtde. Teórica representa uma quantidade inteira, então vamos converter os valores para int (inteiro). Antes de converter, é importante conferir se não há valores vazios ou decimais inesperados.

```python
df_principal['Qtde. Teórica'] = pd.to_numeric(df_principal['Qtde. Teórica'], errors='raise').astype('Int64')
df_principal
```

A coluna Qtde. Teórica tem acento e espaço. O pandas aceita isso, mas vou renomear para ficar mais fácil de usar depois e evitar b.o.

```python
df_principal = df_principal.rename(columns={'Qtde. Teórica':'Qtd_teorica'}).copy()
df_principal
```

Agora a gnt vai fazer oq: pegar a coluna Variacao_rs como base e, se o valor for maior q 0, colocar "Subiu"; se for menor, "Desceu"; se não, "Estável".

.apply aplica uma função aos valores dessa coluna.

lambda cria uma função pequenininha, sem nome, que aqui é usada em cada valor da coluna.

```python
df_principal['Resultado'] = df_principal['Variacao_rs'].apply(lambda x: 'Subiu' if x > 0 else ('Desceu' if x < 0 else 'Estável'))
df_principal
```

Agora a gente vai pegar o ativo e associar com outra planilha para descobrir o nome da empresa e para isso vamos usar novamente o .merge

e oq em cima a gente fez primeiro isso e dps fez o drop, dessa vez já vamos fazer o drop junto

```python
df_principal = df_principal.merge(df_Ticker, left_on ='Ativo', right_on='Ticker', how='left')
df_principal = df_principal.drop(columns=['Ticker'])
df_principal
```

Agora vamos puxar a aba chatgpt para podermos pegar o segmento e a idade dessas empresas, usando o merge novamente e o drop

```python
df_principal = df_principal.merge(df_CHATGPT, left_on='Nome', right_on='Nome da Empresa', how='left')
df_principal =  df_principal.drop(columns=['Nome da Empresa'])
df_principal
```

Agora para gente poder definir as empresas cm mais de 100 anos, menos de 50 anos e entre 100 e 50 anos

```python
df_principal['Cat_idade'] = df_principal['Idade em Anos'].apply(lambda x: 'Mais de 100 anos' if x > 100 else ('Menos que 50' if x < 50 else 'Entre 50 e 100'))
df_principal
```

Agora a gente começa a fazer as análises

```python

# Calculando as estatísticas
maior = df_principal['Variacao_rs'].max()
menor = df_principal['Variacao_rs'].min()
media = df_principal['Variacao_rs'].mean()
media_subiu = df_principal.loc[df_principal['Resultado'] == 'Subiu', 'Variacao_rs'].mean()
media_desceu = df_principal.loc[df_principal['Resultado'] == 'Desceu', 'Variacao_rs'].mean()

# Imprimindo os resultados
print("Maior\tR$ {:.2f}".format(maior))
print("Menor\tR$ {:.2f}".format(menor))
print("Média\tR$ {:.2f}".format(media))
print("Média de quem subiu\tR$ {:.2f}".format(media_subiu))
print("Média de quem desceu\tR$ {:.2f}".format(media_desceu))
```

Criando um novo df só com os casos que subiram

```python
df_principal_subiu = df_principal.loc[df_principal['Resultado'] == 'Subiu']
df_principal_subiu
```

Agora vamos agrupar tudo por segmento (analisando somente os q subiram).

.groupby separa os dados em grupos; o .sum é que soma os valores dentro de cada grupo.

Criamos um novo df só para essa análise.

```python
df_analise_segmento = df_principal_subiu.groupby('Segmento')['Variacao_rs'].sum().reset_index()
df_analise_segmento
```

Agora vamos analisar não só entre os que subiram, e sim geral

```python
df_analise_saldo = df_principal.groupby('Resultado')['Variacao_rs'].sum().reset_index()
df_analise_saldo
```

Agora com esses df’s vamos criar os gráficos.

Vamos usar a biblioteca Plotly Express, que importamos lá em cima junto do pandas.

O apelido para o Plotly Express é px.

fig é a variável que guarda a figura.
.bar é pq estamos criando um gráfico de barras.

x = eixo horizontal
y = eixo vertical

text é para indicar os valores que vão aparecer como rótulos nas barras.

fig.show() é para mostrar o gráfico.

```python
fig = px.bar(df_analise_saldo, x='Resultado', y='Variacao_rs', text='Variacao_rs', title='Variação reais por resultado')
fig.show()
```

```python
fig = px.pie(df_analise_segmento, names='Segmento', values='Variacao_rs', title='Variação Reais por Segmento')
fig.show()
```

```python
df_analise_cat_idade = df_principal.groupby('Cat_idade')['Variacao_rs'].sum().reset_index()
df_analise_cat_idade

fig = px.bar(df_analise_cat_idade, x='Cat_idade', y='Variacao_rs', text='Variacao_rs', title='Variação Reais por Categoria de Idade')
fig.show()
```
